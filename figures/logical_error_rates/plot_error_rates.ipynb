# Final two-panel logical-error-rate figure

Generate only `stacked_error_rate_fits.pdf`: Cornucopia in panel a, and
Cornucopia, bivariate-bicycle, and surface codes in panel b. The stored
X/Z-averaged points and fitted coefficients are included; raw shot files
and decoder packages are not needed.

For matched X- and Z-basis runs, the stored rate is
$$P_B=(P_{B,X}+P_{B,Z})/2,\qquad p_L=1-(1-P_B)^{1/(kT)}.$$
Here $k$ is the logical-qubit count and $T$ is the number of syndrome cycles.
This effective rate uses a factorization convention. Error bars use the
stored counting approximation $p_L/\sqrt{N_{\mathrm{fail},X}+N_{\mathrm{fail},Z}}$.
